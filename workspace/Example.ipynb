Import modules from the standard library and packages.

```python
import datetime
import random


from pytz import timezone
from math import sin, cos, pi
```

The `GF_VARS` is a dict-like variable defined by the Jupyter datasource in the query editor. These are typically the dashboard variables or constants defined within the Grafana dashboard environment. A `try-except` block is used to enable default values and make this notebook work outside of the Jupyter datasource context.

```python
try:
    range_max = GF_VARS.int("RANGE_MAX", 1000)
    freq0 = GF_VARS.float("FREQ0", 7)
    freq1 = GF_VARS.float("FREQ1", 13)
    tz = GF_VARS.str("TZ", "America/Los_Angeles")
except NameError:
    range_max = 1000
    freq0 = 7.0
    freq1 = 13.0
    tz = "America/Los_Angeles"

print(f"{range_max=}")
print(f"{freq0=}")
print(f"{freq1=}")
print(f"{tz=}")
```

The `create_frame` function is defined in the Jupyter datasource. This avoids an error if the notebook is executed outside of the datasource context, similar to the `GF_VARS` variable.

```python
try:
    create_frame
except NameError:
    from IPython.display import JSON
    from typing import Any, Dict, List
    
    def create_frame(name: str, data: List[Dict[str, Any]]):
        return JSON([{"name": name, "data": data}])
```

Create a single time series with a sine and cosine waveforms with different frequencies based on either the variables passed from the Grafana dashboard in the query editor via the `GF_VARS` variable or the default values if running outside of the datasource context.

```python
base = datetime.datetime.now(timezone(tz))

timestamps = [(base - datetime.timedelta(seconds=i)).isoformat() for i in range(range_max)]
sine_wave = [sin(i/range_max * freq0 * pi) for i in range(range_max)]
cosine_wave = [cos(i/range_max * freq1 * pi) for i in range(range_max)]

create_frame("frame", [
  {"name": "time", "values": timestamps},
  {"name": "sine", "values": sine_wave},
  {"name": "cosine", "values": cosine_wave}
])
```
